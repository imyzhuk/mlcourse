8. Among whom is the proportion of those who earn a lot (>50K) greater: married or single men (marital-status feature)? Consider as married those who have a marital-status starting with Married (Married-civ-spouse, Married-spouse-absent or Married-AF-spouse), the rest are considered single.

```python
mask_rich = df["salary"] == ">50K"
mask_men = df["sex"] == "Male"
the_rich = df[mask_rich & mask_men]
married_series = the_rich["marital-status"].apply(lambda status: 1 if status in ["Married-civ-spouse", "Married-spouse-absent", "Married-AF-spouse"] else 0)

married_count = married_series.sum()
non_married_count = married_series.shape[0] - married_count
(married_series.shape[0], married_count, non_married_count)
```

```python
# single men
df[
    (df["sex"] == "Male")
    & ~(df["marital-status"].str.startswith("Married"))
]["salary"].value_counts(normalize=True)
```

9. What is the maximum number of hours a person works per week (hours-per-week feature)? How many people work such a number of hours, and what is the percentage of those who earn a lot (>50K) among them?

```python
max_hours = df["hours-per-week"].max()

workaholics_mask = df["hours-per-week"] == max_hours
workaholics_df = df[workaholics_mask]
workaholics_count = workaholics_df.shape[0]

rich_workaholics_count = (workaholics_df["salary"] == ">50K").sum()
rich_workaholics_percentage = rich_workaholics_count / workoholics_count * 100

print(max_hours, workaholics_count, np.round(rich_workaholics_percentage, 1))
```
