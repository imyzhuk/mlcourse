## Group by

```python
type(df.groupby("State"))
```

```python
for state, sub_dataframe in df.groupby("State"):
    print(state, sum(sub_dataframe["Total day minutes"]))
```

```python
df.groupby("State")["Total day minutes"].sum()
```

### Сводная таблица

```python
pd.crosstab(df["State"], df["Churn"])
```

```python
df.pivot_table(
    ["Total day calls", "Total eve calls", "Total night calls"],
    ["Area code"],
    aggfunc="mean",
)
```

```python
df["Total inner calls"] = df["Total day calls"] + df["Total eve calls"] + df["Total night calls"]
```

```python
df[["Total inner calls", "Total day calls", "Total eve calls", "Total night calls"]].head()
```

```python
df["is_talkative"] = (df["Total inner calls"] > 300).astype("int")
```

```python
df[["Total inner calls", "is_talkative"]].head()
```

### Еще методы
apply - применение ко всем элементам столбца функции,
map - применение ко всем элементам стобца значения из словаря

```python
df["is_talkative"] = df["is_talkative"].apply(lambda flag: "talkative" if flag else "non-talkative")
```

```python
df[["Total inner calls", "is_talkative"]].head()
```

```python
df["int_is_talkative"] = df["is_talkative"].map({"talkative": 1, "non-talkative": 0})
df[["is_talkative", "int_is_talkative"]].head()
```
